```python
from verifier import IBPVerifier
```

```python
verif = IBPVerifier("models/toy_model.onnx")
```

```python
import numpy as np
```

```python
x = np.array([0.5], dtype=np.float32)
eps = 0.1
```

```python
lower_bound, upper_bound = verif.forward(x, eps)

print("Borne inférieure (Lower) :", lower_bound)
print("Borne supérieure (Upper) :", upper_bound)
```
